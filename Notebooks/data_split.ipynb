```python
import os
from sklearn.model_selection import train_test_split
import shutil
from tqdm.auto import tqdm
```

```python
%cd ..
```

```python
src_dir = r"Data\imbalanced"
dest_dir = r"Data\new_data"
```

```python
classes = os.listdir(src_dir)
```

```python
for cls in tqdm(classes):
    file_path = os.path.join(src_dir, cls)
    images = os.listdir(file_path)

    train_split, test_split = train_test_split(images, test_size=0.15)
    ratio = 0.15/0.85
    train_split, val_split = train_test_split(train_split, test_size=ratio)

    for img in train_split:
        os.makedirs(os.path.join(dest_dir, "train", cls), exist_ok=True)
        shutil.move(os.path.join(file_path, img), os.path.join(dest_dir, "train", cls, img))
    for img in test_split:
        os.makedirs(os.path.join(dest_dir, "test", cls), exist_ok=True)
        shutil.move(os.path.join(file_path, img), os.path.join(dest_dir, "test", cls, img))
    for img in val_split:
        os.makedirs(os.path.join(dest_dir, "val", cls), exist_ok=True)
        shutil.move(os.path.join(file_path, img), os.path.join(dest_dir, "val", cls, img))
```
