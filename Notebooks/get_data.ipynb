```python
import os
import shutil

# creates ./kaggle file in the home dir
kaggle_dir = os.path.expanduser("~/.kaggle")
os.makedirs(kaggle_dir, exist_ok=True)

# Set permissions (important for Linux/Mac)
if os.name != 'nt': # 'nt' is Windows
    os.chmod(kaggle_dir, 0o700)

source_file = "kaggle.json" # should be in the project dir
dest_file = os.path.join(kaggle_dir, "kaggle.json")

# Check if kaggle.json is in your project folder
if os.path.exists(source_file):
    # Move the file
    shutil.copyfile(source_file, dest_file)

    # Set file permissions (important for Linux/Mac)
    if os.name != 'nt':
        os.chmod(dest_file, 0o600)
    print(f"Successfully moved {source_file} to {dest_file}")

else:
    print(f"Error: '{source_file}' not found in your project directory.")
    print("Please download it from Kaggle and place it in the same folder as your notebook.")
```
